```python
from prepare.dataset import get_datasets

df = get_datasets(submission_round=1, balance=True, target_per_class=5000)
```

```python
import matplotlib.pyplot as plt

counts = df.groupby("Label").size().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(counts.index, counts.values, color="#4C72B0", edgecolor="white")
ax.bar_label(bars, fmt="{:,.0f}", padding=4, fontsize=9)
ax.set_title("Amostras por label", fontweight="bold")
# ax.set_xlabel("Label")
# ax.set_ylabel("Contagem")
ax.set_ylim(0, counts.max() * 1.12)
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
