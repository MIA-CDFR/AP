```python
import pandas as pd

from pathlib import Path

from models.transformer.transform import TransformModel

project_root = Path().resolve().parents[1]
```

```python
model_path = project_root / "models" / "transformer-fp16.pth"
model = TransformModel.load(model_path)
```

```python
dataset_path = project_root / "src" / "data" / "subm1_labels_revealed.csv"

df = pd.read_csv(dataset_path, sep=";")
```
