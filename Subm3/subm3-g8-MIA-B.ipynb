```python
import sys
import os

sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src')) 
```

```python
import pandas as pd

from pathlib import Path

from models.transformer.model import TransformModel

project_root = Path().resolve().parent

model_path = project_root / "models" / "transformer.pt"
model = TransformModel.load(model_path)
```

```python
dataset_path = project_root / "src" / "data" / "subm2.csv"

df = pd.read_csv(dataset_path, sep=";")
```

```python
X = df["Text"].tolist()

predicted_labels = model.predict(X)

df["Labels"] = predicted_labels
```

```python
df.to_csv("./subm2-g8-MIA-B.csv", index=False, sep=";")
```
