```python
import pandas as pd

from pathlib import Path

from models.pytorch.model import PyTorchModel

project_root = Path().resolve().parent

model_path = project_root / "models" / "pytorch-dnn.pt"
model = PyTorchModel.load(model_path)

project_root = Path().resolve().parent
```
