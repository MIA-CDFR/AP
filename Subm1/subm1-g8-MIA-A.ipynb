```python
import sys
import os

sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src')) 
```

```python
import pandas as pd

from pathlib import Path

from dnn.prepare.model import Model

loaded_model = Model.load("../src/dnn/rnn-model.pkl")

project_root = Path().resolve().parent

dataset_path = project_root / "src" / "dnn" / "data" / "subm1.csv"
```

```python
df = pd.read_csv(dataset_path, sep=";")
```

```python
X = df["Text"].tolist()

predicted_labels = loaded_model.predict(X)
```

```python
df["Labels"] = predicted_labels
```

```python
df.to_csv("./subm1-g8-MIA-A.csv", index=False, sep=";")
```
